**Objective:** Explore the sensor telemetry dataset, inspect statistical distributions, understand physical sensor correlations, and examine how sensor signals shift across operating conditions (`0 = NORMAL`, `1 = WARNING`, `2 = FAULT`).

```python
import sys
from pathlib import Path

# Add project root to sys.path
PROJECT_ROOT = Path("..").resolve()
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from src.config import RAW_DATA_DIR, PLOTS_DIR, SENSOR_FEATURES, STATUS_MAP
from src.data_loader import load_and_validate
```

### 1. Load and Validate Dataset

```python
csv_path = RAW_DATA_DIR / "synthetic_machine_data.csv"
df, report = load_and_validate(csv_path)
report.print_summary()
df.head()
```

### 2. Dataset Statistical Summary & Class Distribution

```python
print("=== Overall Descriptive Statistics ===")
display(df[SENSOR_FEATURES].describe())

print("\n=== Class Breakdown ===")
for code, name in STATUS_MAP.items():
    count = (df['status'] == code).sum()
    pct = (count / len(df)) * 100
    print(f"[{code}] {name:8s}: {count:4d} rows ({pct:.1f}%)")
```

### 3. Sensor Means by Operating Condition
Notice how RPM drops while temperature, current, and vibration increase as the condition worsens.

```python
grouped = df.groupby("status")[SENSOR_FEATURES].mean().reset_index()
grouped["condition_name"] = grouped["status"].map(STATUS_MAP)
display(grouped)
```
